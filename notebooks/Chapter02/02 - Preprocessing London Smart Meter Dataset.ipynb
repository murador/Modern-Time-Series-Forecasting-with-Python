### Converting the half hourly block level dataset into a time series data 

Let's pick one block and see how we can transform the data. 
NOTA: Convertiamo il dataset, che contiene misurazioni aggregate per blocchi di 30 minuti, in una vera serie temporale. Prendiamo un singolo blocco e vediamo come riorganizzarne i dati.

```python
block_1 = pd.read_csv(block_data_path/"block_0.csv", parse_dates=False)

block_1['day'] = pd.to_datetime(block_1['day'], yearfirst=True)

block_1.head()
```

```python
#Check End Dates of all time series
block_1.groupby("LCLid")['day'].max().sample(5)
```

IL codice seguente legge tutti i file CSV in block_data_path, converte la colonna day in datetime e mantiene il massimo tra tutte le date trovate. Alla fine stampa la data più recente presente in tutti i file.

```python
max_date = None
for f in tqdm(block_data_path.glob("*.csv")):
    df = pd.read_csv(f, parse_dates=False)
    df['day'] = pd.to_datetime(df['day'], yearfirst=True)
    if max_date is None:
        max_date = df['day'].max()
    else:
        if df['day'].max()>max_date:
            max_date = df['day'].max()
print(f"Max Date across all blocks: {max_date}")
del df
```

Il codice trasforma `block_1` da un formato “wide” con colonne `hh_0`, `hh_1`, … a un formato “long” in cui ogni riga rappresenta un singolo blocco orario.

Passaggi:

1. `block_1.set_index(['LCLid', "day"])`
   - imposta le colonne `LCLid` e `day` come indice temporaneo.
   - il resto delle colonne (`hh_0`, `hh_1`, …) rimane come colonne da riorganizzare.

2. `.stack()`
   - prende tutte le colonne rimanenti e le “impila” lungo l’indice.
   - ogni colonna `hh_x` diventa una riga in più, con un nuovo livello di indice che identifica il nome della colonna.
   - il risultato è una `Series` con indice multi-livello: (`LCLid`, `day`, `hour_block`).

3. `.reset_index()`
   - converte i livelli dell’indice in colonne normali.
   - quindi `LCLid` e `day` tornano ad essere colonne, e il terzo livello di indice diventa una colonna chiamata `level_2`.

4. `.rename(columns={"level_2": "hour_block", 0: "energy_consumption"})`
   - rinomina la colonna generata da `stack()`:
     - `level_2` → `hour_block`
     - la colonna dei valori impilati `0` → `energy_consumption`

A questo punto `block_1` ha righe del tipo:
- `LCLid`
- `day`
- `hour_block` (es. `hh_0`, `hh_1`, …)
- `energy_consumption`

5. `block_1['offset'] = block_1['hour_block'].str.replace("hh_", "").astype(int)`
   - `str.replace("hh_", "")`: sul `Series` di stringhe `hour_block`, rimuove il prefisso `hh_`.
   - `.astype(int)`: converte il risultato in numeri interi.
   - otteniamo `offset` = 0, 1, 2, …, cioè il blocco orario numerico.

6. `block_1.head()`
   - mostra le prime 5 righe del DataFrame risultante.
   - è utile per verificare il risultato della trasformazione.

In sintesi:
- `set_index` cambia l’indice del DataFrame.
- `stack` trasforma le colonne rimanenti in righe.
- `reset_index` riporta l’indice in colonne.
- `rename` rinomina colonne.
- `str.replace` modifica stringhe in un `Series`.
- `astype(int)` converte il tipo in intero.
- `head()` mostra le prime righe.

```python
#Reshaping the dataframe into the long form with hour blocks along the rows
block_1 = block_1.set_index(['LCLid', "day"]).stack().reset_index().rename(columns={"level_2": "hour_block", 0: "energy_consumption"})
#Creating a numerical hourblock column
block_1['offset'] = block_1['hour_block'].str.replace("hh_", "").astype(int)

block_1.head()
```

#### Compact Form

```python
def preprocess_compact(x):
    start_date = x['day'].min()
    name = x['LCLid'].unique()[0]
    ### Fill missing dates with NaN ###
    # Create a date range from  min to max
    dr = pd.date_range(start=x['day'].min(), end=max_date, freq="1D")
    # Add hh_0 to hh_47 to columns and with some unstack magic recreating date-hh_x combinations
    dr = pd.DataFrame(columns=[f"hh_{i}" for i in range(48)], index=dr).unstack().reset_index()
    # renaming the columns
    dr.columns = ["hour_block", "day", "_"]
    # left merging the dataframe to the standard dataframe
    # now the missing values will be left as NaN
    dr = dr.merge(x, on=['hour_block','day'], how='left')
    # sorting the rows
    dr.sort_values(['day',"offset"], inplace=True)
    # extracting the timeseries array
    ts = dr['energy_consumption'].values
    len_ts = len(ts)
    return start_date, name, ts, len_ts
```

```python
def load_process_block_compact(block_df, freq="30min", ts_identifier="series_name", value_name="series_value"):
    grps = block_df.groupby('LCLid')
    all_series = []
    all_start_dates = []
    all_names = []
    all_data = {}
    all_len = []
    for idx, df in tqdm(grps, leave=False):
        start_date, name, ts, len_ts = preprocess_compact(df)
        all_series.append(ts)
        all_start_dates.append(start_date)
        all_names.append(name)
        all_len.append(len_ts)

    all_data[ts_identifier] = all_names
    all_data['start_timestamp'] = all_start_dates
    all_data['frequency'] = freq
    all_data[value_name] = all_series
    all_data['series_length'] = all_len
    return pd.DataFrame(all_data)

block1_compact = load_process_block_compact(block_1, freq="30min", ts_identifier="LCLid", value_name="energy_consumption")
```

#### Expanded Form

```python
def preprocess_expanded(x):
    start_date = x['day'].min()
    ### Fill missing dates with NaN ###
    # Create a date range from  min to max
    dr = pd.date_range(start=x['day'].min(), end=x['day'].max(), freq="1D")
    # Add hh_0 to hh_47 to columns and with some unstack magic recreating date-hh_x combinations
    dr = pd.DataFrame(columns=[f"hh_{i}" for i in range(48)], index=dr).unstack().reset_index()
    # renaming the columns
    dr.columns = ["hour_block", "day", "_"]
    # left merging the dataframe to the standard dataframe
    # now the missing values will be left as NaN
    dr = dr.merge(x, on=['hour_block','day'], how='left')
    dr['series_length'] = len(dr)
    return dr
```

```python
def load_process_block_expanded(block_df, freq="30min"):
    grps = block_df.groupby('LCLid')
    all_series = []
    for idx, df in tqdm(grps, leave=False):
        ts = preprocess_expanded(df)
        all_series.append(ts)

    block_df = pd.concat(all_series)
    # Recreate Offset because there would be null rows now
    block_df['offset'] = block_df['hour_block'].str.replace("hh_", "").astype(int)
    # Creating a datetime column with the date | Will take some time because operation is not vectorized
    block_df['timestamp'] = block_df['day'] + block_df['offset']*30*pd.offsets.Minute()
    block_df['frequency'] = freq
    block_df.sort_values(["LCLid","timestamp"], inplace=True)
    block_df.drop(columns=["_", "hour_block", "offset", "day"], inplace=True)
    return block_df
#     del all_series
block1_expanded = load_process_block_expanded(block_1, freq="30min")
```

#### Reading and combining all the block data into a single dataframe

```python
block_df_l = []
for file in tqdm(sorted(list(block_data_path.glob("*.csv"))), desc="Processing Blocks.."):
    block_df = pd.read_csv(file, parse_dates=False)
    block_df['day'] = pd.to_datetime(block_df['day'], yearfirst=True)
    # Taking only from 2012-01-01
    block_df = block_df.loc[block_df['day']>="2012-01-01"]
    #Reshaping the dataframe into the long form with hour blocks along the rows
    block_df = block_df.set_index(['LCLid', "day"]).stack().reset_index().rename(columns={"level_2": "hour_block", 0: "energy_consumption"})
    #Creating a numerical hourblock column
    block_df['offset'] = block_df['hour_block'].str.replace("hh_", "").astype(int)
    block_df_l.append(load_process_block_compact(block_df, freq="30min", ts_identifier="LCLid", value_name="energy_consumption"))
```

```python
hhblock_df = pd.concat(block_df_l)
del block_df_l
display(hhblock_df.memory_usage(deep=True))
print(f"Total: {hhblock_df.memory_usage(deep=True).sum()/1024**2} MB")
```

#### Weather and Bank Holidays

```python
bank_holidays = pd.read_csv("C:\\Users\\39340\\Documents\\PythonProject\\Murador_repo\\Modern-Time-Series-Forecasting-with-Python\\data\\london_smart_meters\\smart-meters-in-london\\uk_bank_holidays.csv", parse_dates=False)
bank_holidays['Bank holidays'] = pd.to_datetime(bank_holidays['Bank holidays'], yearfirst=True)
bank_holidays.set_index("Bank holidays", inplace=True)
bank_holidays.head()
```

```python
# Reindex on standard date range
# Steps:
# 1) resample("30min").asfreq() -> riallinea l'indice temporale 
# a intervalli di 30 minuti senza aggregare (crea NaN per i timestamps mancanti).
# 2) groupby(bank_holidays.index.date).ffill().fillna("NO_HOLIDAY") -> raggruppa per giorno,
#  propaga in avanti i valori di festività all'interno di ogni giorno e sostituisce i NaN rimanenti con "NO_HOLIDAY".
# 3) Imposta il nome dell'indice su "datetime" e mostra le prime righe con head().
bank_holidays = bank_holidays.resample("30min").asfreq()
bank_holidays = bank_holidays.groupby(bank_holidays.index.date).ffill().fillna("NO_HOLIDAY")
bank_holidays.index.name="datetime"
bank_holidays.head()
```

Cosa significa "New Year's Day (substitute day)":

indica il giorno festivo osservato (observed/substitute holiday). Quando la data “ufficiale”
 della festività cade in un weekend, l’osservanza può essere spostata a un giorno lavorativo 
 (il “substitute day”) — il CSV riporta il nome della festività osservata per quella data.

```python
weather_hourly = pd.read_csv("C:\\Users\\39340\\Documents\\PythonProject\\Murador_repo\\Modern-Time-Series-Forecasting-with-Python\\data\\london_smart_meters\\smart-meters-in-london\\weather_hourly_darksky.csv", parse_dates=False)
weather_hourly['time'] = pd.to_datetime(weather_hourly['time'], yearfirst=True)
weather_hourly.set_index("time", inplace=True)
weather_hourly.head()
```

```python
#Resampling at 30min and forward fill
#forward fill is used to fill missing values in the
# resampled data by propagating the last valid observation forward to the next valid one.
# commento in italiano: il codice seguente serve a riallineare i dati meteo a intervalli di 30 minuti, 
# riempiendo eventuali valori mancanti con l'ultimo valore valido disponibile.
# Nota che la pressione il vento e la temperatura sono misurati ogni ora, 
# quindi il forward fill è necessario per ottenere valori a intervalli di 30 minuti.
weather_hourly = weather_hourly.resample("30min").ffill()
weather_hourly.head()
```

```python
# La procedura seguente mappa le festività e i dati meteo a 
# ciascuna serie temporale nel dataframe hhblock_df. Per fare ciò, crea un intervallo di date
#  a partire dal timestamp iniziale della serie temporale e lo utilizza 
# per creare un dataframe standard. Successivamente, unisce questo dataframe con i dati 
# delle festività e del meteo, riempiendo eventuali valori mancanti con "NO_HOLIDAY"
#  per le festività. Infine, assegna i valori delle festività e del meteo alla riga 
# corrispondente della serie temporale. In questo caso i record sono mappati a intervalli di 30 minuti,
#  quindi la frequenza è impostata su "30min", quindi ogni record è un intervallo di 30 minuti del giorno. 
# La funzione restituisce la riga aggiornata con le informazioni sulle festività e i dati meteo corrispondenti a ciascun giorno della serie temporale.
# Ogni riga rappresenta una serie temporale di consumo energetico per un determinato LCLid,
# e la funzione restituisce la riga aggiornata con le informazioni sulle festività e i dati meteo corrispondenti a ciascun timestamp della serie temporale.
# 
def map_weather_holidays(row):
    date_range = pd.date_range(row['start_timestamp'], periods=row['series_length'], freq=row['frequency'])
    std_df = pd.DataFrame(index=date_range)
    #Filling Na iwth NO_HOLIDAY cause rows before earliers holiday will be NaN
    holidays = std_df.join(bank_holidays, how="left").fillna("NO_HOLIDAY")
    weather = std_df.join(weather_hourly, how='left')
    assert len(holidays)==row['series_length'], "Length of holidays should be same as series length"
    assert len(weather)==row['series_length'], "Length of weather should be same as series length"
    row['holidays'] = holidays['Type'].values
    for col in weather:
        row[col] = weather[col].values
    return row
```

```python
hhblock_df = hhblock_df.progress_apply(map_weather_holidays, axis=1)

hhblock_df.head(5)
```
